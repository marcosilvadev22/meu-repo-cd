```python
import pandas as pd

sales = pd.DataFrame(
    {
        "product": ["Caneca", "Camiseta", "Caneca", "Adesivo", "Camiseta"],
        "quantity": [2, 1, 4, 10, 3],
        "unit_price": [35.0, 79.9, 35.0, 5.5, 79.9],
        "country": ["Brasil", "Brasil", "Portugal", "Brasil", "Argentina"],
    }
)

sales["revenue"] = sales["quantity"] * sales["unit_price"]
brazil_filter = sales["country"].eq("Brasil")
high_revenue_filter = sales["revenue"].gt(50)
sales.loc[brazil_filter & high_revenue_filter]
```

CARREGAMENTO DO DATASET REAL ONLINE RETAIL

```python
from pathlib import Path
import pandas as pd

data_path = Path("..") / "data" / "online_retail.xlsx"

column_names = {
    "InvoiceNo": "invoice_no",
    "StockCode": "stock_code",
    "Description": "description",
    "Quantity": "quantity",
    "InvoiceDate": "invoice_date",
    "UnitPrice": "unit_price",
    "CustomerID": "customer_id",
    "Country": "country",
}

retail = pd.read_excel(data_path).rename(columns=column_names)
retail.head()
```

INSPEÇÃO INICIAL

```python
display(retail.head())
display(retail.tail())

print("Shape:", retail.shape)
retail.info()

print("\nValores ausentes por coluna:")
print(retail.isna().sum().sort_values(ascending=False))

display(retail.describe())

print("\nTop 10 países:")
print(retail["country"].value_counts().head(10))
print("\nTotal de países únicos:", retail["country"].nunique())
```

EXEMPLOS DE FILTRAGEM E DIAGNÓSTICO DE QUALIDADE

```python
germany_filter = retail["country"].eq("Germany")
germany_sales = retail.loc[germany_filter].copy()
print("Vendas na Alemanha:", germany_sales.shape)

expensive_outside_uk_filter = (
    retail["unit_price"].gt(100)
    & retail["country"].ne("United Kingdom")
)
result_columns = ["invoice_no", "description", "unit_price", "country"]
expensive_outside_uk = retail.loc[expensive_outside_uk_filter, result_columns].copy()
display(expensive_outside_uk.head())

negative_quantity_filter = retail["quantity"].lt(0)
negative_quantity = retail.loc[negative_quantity_filter].copy()
print(f"{len(negative_quantity):,} linhas com quantidade negativa")

negative_invoices = retail.loc[negative_quantity_filter, "invoice_no"]
starts_with_c = negative_invoices.str.startswith("C", na=False)
print(f"Começam com C: {starts_with_c.sum():,} | Total de negativas: {len(negative_invoices):,}")

zero_price_filter = retail["unit_price"].eq(0)
zero_price = retail.loc[zero_price_filter].copy()
print(f"{len(zero_price):,} linhas com preço zero")

duplicate_count = retail.duplicated().sum()
print(f"{duplicate_count:,} linhas duplicadas")
```

## Ficha de Reconhecimento

* **Tamanho:** O dataset possui **541.909 observações (linhas)** e **8 variáveis (colunas)**, ocupando cerca de 33,1+ MB em memória[cite: 38].
* **Classificação das variáveis:**
  * `invoice_no` (object): Categórica nominal (identificador da nota fiscal; prefixo `C` indica cancelamento).
  * `stock_code` (object): Categórica nominal (código identificador do produto).
  * `description` (object): Categórica nominal (descrição textual do produto).
  * `quantity` (int64): Numérica discreta (quantidade de unidades do item na transação).
  * `invoice_date` (datetime64[ns]): Temporal (data e hora exatas da transação).
  * `unit_price` (float64): Numérica contínua (preço unitário em libras esterlinas).
  * `customer_id` (float64 no pandas, mas conceitualmente **categórica nominal**): Código identificador único do cliente (aparece como decimal devido à presença de valores nulos `NaN`).
  * `country` (object): Categórica nominal (país de residência do cliente — 38 países distintos, com ampla predominância do Reino Unido).
* **Valores ausentes (`NaN`):**
  * `customer_id`: **135.080** valores ausentes (~24,9% da base).
  * `description`: **1.454** valores ausentes (~0,27% da base).
  * As demais 6 colunas não possuem valores ausentes.
* **Três problemas de qualidade diagnosticados (sem alteração nesta etapa):**
  1. **Quantidades negativas (`10.624` linhas):** Associadas majoritariamente a notas de cancelamento/devolução (prefixo `C` em `invoice_no`), além de possíveis ajustes de estoque.
  2. **Preços unitários zerados (`2.515` linhas) e negativos (`2` linhas, como `-11062.06`):** Itens sem valor comercial registrado, frequentemente sem `description` e sem `customer_id` (ajustes contábeis/perdas).
  3. **Registros duplicados (`5.268` linhas):** Linhas idênticas em todas as colunas, que precisam ser investigadas antes de decidir se são falhas de sistema ou compras repetidas legítimas no mesmo minuto.

Pergunta 1:  O pandas carregou customer_id como float64 porque essa coluna contém 135.080 valores ausentes (NaN). Nas versões tradicionais do NumPy/pandas, o tipo inteiro padrão (int64) não suporta representar NaN (que é um valor especial de ponto flutuante); por isso, sempre que uma coluna numérica inteira tem pelo menos um valor nulo, o pandas a converte automaticamente para float64 (adicionando .0 aos códigos).
